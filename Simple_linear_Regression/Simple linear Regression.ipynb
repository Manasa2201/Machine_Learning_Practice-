```python
import pandas as pd
df = pd.read_csv("income_dataset_2025.csv")
df
```

```python
plt.figure(figsize = (5,4))
plt.scatter(df.Year, df.Income)
```

```python
# Searching the input feature and output label

x = df.drop("Income", axis = 1)  # x is input feature
y = df.Income  # Target / Output label
```

```python
# Split data into training and testing
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.20, random_state = 42 )

print(len(x_train))
CCC
```

```python
# Train the modelw
from sklearn import linear_model

reg_obj = linear_model.LinearRegression()     # Creating SLR model Object

reg_obj.fit(x_train,y_train)    # fit() function is finding the best fit line
```

```python
reg_obj.predict([[2030]])
```

# Performance of ml models

```python
reg_obj.score(x_test, y_test)
```

```python
y_predict = reg_obj.predict(x_test)

# Compare thr actual vs Prediction
for actual, pred in zip(y_test, y_predict):
    print(actual, " : ", pred)
```

```python
# Plotting the best fit line
import pandas as pd
df = pd.read_csv("income_dataset_2025.csv")
df
import matplotlib.pyplot as plt

plt.scatter(df.Year, df.Income)
plt.plot(df.Year, reg_obj.predict(df[['Year']]), color = 'red')
```
