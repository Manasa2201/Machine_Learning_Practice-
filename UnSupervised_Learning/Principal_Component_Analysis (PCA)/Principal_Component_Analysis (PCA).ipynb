# Principal Components Analysis (PCA)

```python
# PCA is used to reduce no of features(Columns)

# PCA isa unsupervised ML Technique, used to reduce the dimensionality of as dataset while preservingthe most imp
# Patterns or relationship between the variables

# PCA can be used for a variety of purposes. including feature selection, finding hidden patterns, and data compression

# PCA is popularly usedto handel curse of dimensionality and multi-collinearity in data

# Curse of Dimensionality = Too many features make data analysis difficult.

# Multicollinearity = Two or more input features are highly correlated.
```

```python
# PCA generate new features (Principal components) from origignal features which have high level of information and less in numbers
# more the variance captured, better the feature

# It gives 2D array to 1D array 

# the line which have more information called Principle components---PCA-1



```

```python
# Libaries

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
```

```python
# Load the dataset

data_obj = load_breast_cancer()
print(dir(data_obj))
```

```python
print(data_obj.DESCR)
```

```python
x = data_obj.data
y = data_obj.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# Apply the PCA

pca_obj = PCA(n_components = 5)
x_pca = pca_obj.fit_transform(x_scaled)
```

```python
x_pca
```

```python
print(pca_obj.explained_variance_ratio_)
```

# Compare x_pca ( 5 Features)  VS x_scaled ( 30 features)

```python
# train the Logistic Regression 

log_obj = LogisticRegression()

log_obj.fit(x_train, y_train)
```

```python
log_obj.score(x_test, y_test)
```
