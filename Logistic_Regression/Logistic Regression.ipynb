```python
# Libararies

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("insurance_logistic_dataset_10k.csv")
df
```

```python
plt.figure(figsize = (5,4))
plt.scatter(df.age, df.bought_insurance)   # The Shape of Logistic Regression is "S"
```

```python
# Split the data into input feature and target

x = df.drop("bought_insurance", axis = 1)
y = df.bought_insurance
```

```python
# divide the data into training and testing

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.8 ,random_state = 42 )
```

```python
# train the Logistic Regression 

log_obj = LogisticRegression()

log_obj.fit(x_train, y_train)
```

```python
log_obj.score(x_test, y_test)
```

```python
# actual vs predicted
# where the actual test output is stored

actual = y_test[0:10]

predicted = log_obj.predict(x_test[0:10])
for act, pred in zip(actual, predicted):
    print(act, " : ", pred)
```

```python
 # OUTPUT all values
y_predict = log_obj.predict(x_test)

# Compare thr actual vs Prediction
for actual, pred in zip(y_test, y_predict):
    print(actual, " : ", pred)
```
