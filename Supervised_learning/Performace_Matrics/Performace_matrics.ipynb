# Performance Matrics

# Confusion Matrix

```python
# Libraries

from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
```

```python
# Load the inbuild dataset

cancer_obj = load_breast_cancer()
dir(cancer_obj)
```

```python
print(cancer_obj.DESCR)
```

```python
# store features inside x variable and target inside y variable

x = cancer_obj.data     # INPUT FEATURES
y = cancer_obj.target   # TARGET VALUES
x
```

```python
y
```

```python
# Split the data into training and testing 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
print(len(x_test))
```

```python
print(len(x_train))
```

```python
# Scaling the data

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# Import Different machine learning algorithms

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

```python
# model object creation

logreg = LogisticRegression()
dtree = DecisionTreeClassifier()
knn = KNeighborsClassifier()
nb = GaussianNB()
svm = SVC()
```

```python
# Training all ml algorithms

logreg.fit(x_train_scaled, y_train)
dtree.fit(x_train_scaled, y_train)
knn.fit(x_train_scaled, y_train)
nb.fit(x_train_scaled, y_train)
svm.fit(x_train_scaled, y_train)
```

```python
# Score

print(logreg.score(x_train_scaled, y_train))
print(dtree.score(x_train_scaled, y_train))
print(knn.score(x_train_scaled, y_train))
print(nb.score(x_train_scaled, y_train))
print(svm.score(x_train_scaled, y_train))
```

```python
# making the predictions

y_pred_logreg = logreg.predict(x_test_scaled)
y_pred_dtree = dtree.predict(x_test_scaled)
y_pred_knn = knn.predict(x_test_scaled)
y_pred_nb = nb.predict(x_test_scaled)
y_pred_svm = svm.predict(x_test_scaled)
```

```python
# y_test is the "actual" target value
import seaborn as sns
import matplotlib.pyplot as plt
# constructur the confusion matrix
plt.figure(figsize = (3,2))
cm_logreg = confusion_matrix(y_test, y_pred_logreg)
sns.heatmap(cm_logreg, annot = True)
plt.xlabel("Predicted values")
plt.ylabel("Actual values")
```

# USING FOR LOOP BETWEEN ML ALGORITHMS

```python

# Create a dictionary of remaining algorithms

models = {
    
    "Decision Tree": dtree,
    "KNN": knn,
    "Naive Bayes": nb,
    "SVM": svm,
    
}

# Plot confusion matrix for each algorithm

plt.figure(figsize=(15, 10))

for i, (name, model) in enumerate(models.items()):

    # Predict values
    y_pred = model.predict(x_test_scaled)

    # Calculate confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    # Create subplot
    plt.subplot(2, 2, i + 1)

    # Plot heatmap
    sns.heatmap(cm_logreg, annot = True)
    plt.title(f'{name} Confusion Matrix')
    plt.xlabel('Predicted Values')
    plt.ylabel('Actual Values')

plt.tight_layout()
plt.show()
```

# ACCURACY

```python
# It tells the percentage total correctly predicted value out of total predictions.

# Accuracy = total correct pred/total pred
# ==> 97/100 ==> 97%

# ACCURACY = TP+TN/ TP+TN+FP+FN
```

# ACCURACY MATRIX

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_svm).ravel()
accuracy =  (TP+TN)/ (TP+TN+FP+FN)
print(accuracy)
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_nb).ravel()
accuracy =  (TP+TN)/ (TP+TN+FP+FN)
print(accuracy)
```

# Precision Matrix

```python
# Out of all the predicted positive values how many are predicted correctly
# precision should be use we are concern about the type - 1 error

# Precision is 0 to 1

# precision is used when a false positive is more imp like in span detection if a mail is "not spam"
# and the model predicted it as a spam, we might miss a very imp mail beacause of this error.
# So FP Value must be minimum to make sure NOT SPAM dont go in spam detection.

# precision = TP/TP+FP   # FP value must be minimum

# POSITIVELY ACCURATE 
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_svm).ravel()
precision =  TP / (TP + FP)  
precision
```

# RECALL OR SENSITIVITY

```python
# Out of all the actual positive values how many are predicted correctly.It is also Known as TRUE POSITIVE RATE(TPR)

# IF FN == > 0 it wll be 1

# Use recall when we have TYPE - 2 Error

# Recall is used when a False negative is more imp like a patient is covid positive but the model detected it as negative,
# it will cause a great impact on healthcare

# RECALL = TP / (TP + FN)
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_svm).ravel()
recall =  TP / (TP + FN)
print(recall)
```

# Specificity and FPR (FALSE POSITIVE RATE)

```python
# Out of all the actual negative values how many are predicted correctly.

# Providing loan to Bad Credit Score candidate(FP) is not a good idea

# Specificity = TN / (TN + FP)

# FPR = FP / (TN + FP)
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_svm).ravel()
Specificity =  TN / (TN + FP)
print(Specificity)
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, y_pred_svm).ravel()
FPR = FP / (TN + FP)
print(FPR)
```
