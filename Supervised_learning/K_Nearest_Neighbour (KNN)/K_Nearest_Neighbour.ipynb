# K Nearest Neighbour (KNN)

```python
# KNN is supervised ML Algorithm.
# KNN mainly used for classification.
# Although KNN can also be used for Regression based Promblems.
# KNN can be used for binary or Multiclass classification.
# KNN is also known as Lazy ML algorithm.
# Maximum calculations are done at the time of prediction.

```

```python
# Working Principle of KNN
# In KNN, we Look for K nearest data points.
# the data points which are in majority decide the category


# Its good idea to take K value as Odd value.
# KNN is slow because this is time consuming to find the nearest data points.
```

```python
# Libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("teleCust1000t.csv")
df
```

```python
# divide the input feature and target

x = df.drop("custcat", axis =1)
y = df.custcat
```

```python
x
```

```python
y
```

```python
# Spliting data into train and test

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
print(len(x_test))
```

```python
print(len(x_train))
```

```python
# Scale the training and testing data
from sklearn.preprocessing import MinMaxScaler

scaled1 = MinMaxScaler()

x_train_scaled = scaled1.fit_transform(x_train)
x_test_scaled = scaled1.transform(x_test)
```

```python
# Train the model

knn_model_obj = KNeighborsClassifier(n_neighbors = 85)
knn_model_obj.fit(x_train_scaled, y_train)
```

```python
# Performance of KNN

knn_model_obj.score(x_test_scaled, y_test)
```

```python
data = [2, 13, 14, 1, 9, 64.0, 4, 5, 0.0, 0, 2]
knn_model_obj.predict([data])
```

# Apply Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
```

```python
NB_obj = GaussianNB()
NB_obj.fit(x_train_scaled, y_train)
```

```python
NB_obj.score(x_test_scaled, y_test)
```

# Apply SVM

```python
from sklearn.svm import SVC
```

```python
svc_obj = SVC()
svc_obj.fit(x_train_scaled, y_train)
```

```python
svc_obj.score(x_test_scaled, y_test)
```

```python
# Libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("Chinky_shopping_regression_dataset.csv")
df
```

```python
# input featurs is x # target is y

x = df.drop("shopping_bill_inr", axis = 1)
y = df.shopping_bill_inr
```

```python
# split the data into training and testing

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
# scale the training and testing data

scaler2 = MinMaxScaler()

x_train_scaled = scaler2.fit_transform(x_train) # for training
x_test_scaled = scaler2.transform(x_test) # for testing
```

```python
# train model

knn_model_obj = KNeighborsRegressor(n_neighbors = 5)
knn_model_obj.fit(x_train_scaled, y_train)
```

```python
# performance

knn_model_obj.score(x_test_scaled, y_test)
```

```python
data = [2, 13, 14, 1, 9, 64.0, 4, 5, 0.0,  2, 2, 1.5, 2.5, 2, 2.3]
knn_model_obj.predict([data])
```
