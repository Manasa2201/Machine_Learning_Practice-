# Decision Tree

```python
# DT is a Supervised Ml Algorithm
# DT is mainly used for classification promblems
# DT can also be used for Regression based promblems
# DT algorithm take time while training.
# DT is fast while making prediction



```

```python
# Features : Age, Weight, Smoker        Target : health issue
```

```python
# Libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine # inbuild dataset
```

```python
# Load the inbuild dataset

wine_obj = load_wine()
dir(wine_obj)
```

```python
print(wine_obj.DESCR)
```

```python
# store features inside x variable and target inside y variable

x = wine_obj.data     # INPUT FEATURES
y = wine_obj.target   # TARGET VALUES
x
```

```python
y
```

```python
# Split the data into training and testing 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
print(len(x_test))
```

```python
print(len(x_train))
```

```python
# Train the DT Tree
# max_depth helping in DT pruning

DT_obj = DecisionTreeClassifier(max_depth = 2)
DT_obj.fit(x_train, y_train)
```

```python
# Performancewwww

DT_obj.score(x_test, y_test)
```

```python
# Plotting the DT

import matplotlib.pyplot as plt
from sklearn.tree import plot_tree


plot_tree(DT_obj, feature_names = wine_obj.feature_names, class_names = wine_obj.target_names,  filled = True)
plt.title("Decision tree for wine data set")
```
