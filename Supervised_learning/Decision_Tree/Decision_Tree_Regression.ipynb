```python
# Libraries

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import fetch_california_housing # inbuild dataset
```

```python
# Load the inbuild dataset

housing_obj = fetch_california_housing()
dir(housing_obj)
```

```python
print(housing_obj.DESCR)
```

```python
# store features inside x variable and target inside y variable

x = housing_obj.data     # INPUT FEATURES
y = housing_obj.target   # TARGET VALUES
x
```

```python
y
```

```python
# Split the data into training and testing 

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
print(len(x_test))
```

```python
print(len(x_train))
```

```python
# Train the DT Tree
# max_depth helping in DT pruning

DT_obj = DecisionTreeRegressor(max_depth = 3)
DT_obj.fit(x_train, y_train)
```

```python
# Performance

DT_obj.score(x_test, y_test)
```

```python
# Plotting the DT

import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize = (0,3))

plot_tree(DT_obj, feature_names = housing_obj.feature_names, class_names = housing_obj.target_names,  filled = True, fontsize = 10)
plt.title("Decision tree for wine data set")
```
